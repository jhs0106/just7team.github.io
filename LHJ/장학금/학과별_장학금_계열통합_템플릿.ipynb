```python
import pandas as pd

df = pd.read_csv("학과별_장학금_21_23_재분류.csv")

# 총 장학금 = 1인당 장학금 × 재학생 수
df['총장학금'] = df['1인당장학금'] * df['재학생']

# 그룹화해서 계산
result = df.groupby(['기준연도', '학교명', '대계열']).agg({'총장학금': 'sum','재학생': 'sum'}).reset_index()

# 대계열별 1인당 장학금 계산
result['대계열_1인당장학금'] = result['총장학금'] / result['재학생']

# 데이터 정리
result = result[['기준연도', '학교명', '대계열', '대계열_1인당장학금']]

result.columns = ['기준연도', '학교명', '대계열', '1인당장학금']

result.to_csv("학과별_장학금_2021_2023_최종.csv", index=False, encoding='utf-8-sig')
```
