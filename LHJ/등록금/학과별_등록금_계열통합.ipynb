```python
import pandas as pd

df = pd.read_csv("학과별_등록금_재분류.csv")

# '등록금' 열이 숫자로 잘 읽혔는지 확인 (쉼표 제거 및 숫자형 변환 필요 시 처리)
df['등록금'] = df['등록금'].astype(str).str.replace(',', '').astype(float)

# 그룹화: 기준연도, 학교명, 대계열 기준 등록금 평균 계산
result = df.groupby(['기준연도', '학교명', '대계열'])['등록금'].mean().reset_index()

result.to_csv("학과별_등록금_최종.csv", index=False, encoding='utf-8-sig')
```
