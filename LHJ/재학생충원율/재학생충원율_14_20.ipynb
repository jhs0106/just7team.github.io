```python
import pandas as pd

df = pd.read_csv("학과별_재학생_충원율_14-20.csv")

df = df.dropna(axis=1, how='all')
df.columns = ['기준연도','학교종류', '학교명',' 대계열', '학생정원', '학생모집정지', '재학생', '재학생충원율']

# 결측값 처리
df['기준연도'] = df['기준연도'].fillna(method='ffill')
df['학교명'] = df['학교명'].fillna(method='ffill')

# 형 변환, 쉼표 제거
df['기준연도'] = df['기준연도'].astype(int)
df['재학생'] = pd.to_numeric(df['재학생'].astype(str).str.replace(',', ''), errors='coerce')
df['재학생충원율'] = pd.to_numeric(df['재학생충원율'].astype(str).str.replace(',', ''), errors='coerce')
df['학생정원'] = pd.to_numeric(df['학생정원'].astype(str).str.replace(',', ''), errors='coerce')
df['학생모집정지'] = pd.to_numeric(df['학생모집정지'].astype(str).str.replace(',', ''), errors='coerce')

# 사이버대학 제거 및 재학생, 재학생충원율, 학생모집, 학생정원이 0 인 행 제거
c1 = df['학교종류'].str.contains('사이버대학', na=False)
c2 = ((df['재학생'] == 0) & (df['재학생충원율'] == 0) & (df['학생모집정지'] == 0) & (df['학생정원'] == 0))
df = df[~(c1 | c2)]
df = df.drop(['학교종류'], axis=1)

# 학생정원 - 학생모집정지 = 실모집인원
df['실모집인원'] = df['학생정원'] - df['학생모집정지']
df = df.drop(['학생정원', '학생모집정지'], axis=1)

# 컬럼, 인덱스 재설정
df.columns = ['기준연도','학교명', '대계열', '재학생', '재학생충원율', '실모집인원']
df = df.reset_index(drop=True)

# 대계열 특수문자 제거
df['대계열'] = df['대계열'].str.replace(r'[ㆍ·\-/]', '', regex=True)

df.to_csv("학과별_재학생충원율_14-20_최종.csv", index=False, encoding='utf-8-sig')
```
