```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams['font.family'] = 'NanumGothic'
plt.rcParams['axes.unicode_minus'] = False
sns.set(style="whitegrid")
```

```
df = pd.read_csv("신입생기준_Zscore_결측치제거_170개학교.csv", encoding="utf-8-sig")
df.shape
```

```
df.isnull().sum()
```

```
numeric_cols = df.select_dtypes(include='number').columns.drop('기준연도', errors='ignore')
df[numeric_cols].hist(bins=20, figsize=(15, 10))
plt.suptitle("수치형 피처 분포", fontsize=16)
plt.show()
```

```
corr = df[numeric_cols].corr()
plt.figure(figsize=(12, 8))
sns.heatmap(corr, annot=True, fmt=".2f", cmap='coolwarm')
plt.title("상관관계 히트맵")
plt.show()
```

```
target = "중도탈락학생(신입생)비율(%)"
for col in numeric_cols:
    if col != target:
        plt.figure()
        sns.scatterplot(data=df, x=col, y=target, hue=df['기준연도'], palette="tab10")
        plt.title(f"{target} vs {col} (기준연도 표시)")
        plt.xlabel(col)
        plt.ylabel(target)
        plt.legend(title="기준연도")
        plt.show()
```
