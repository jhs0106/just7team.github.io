파일 : 혁진씨 계열별 전처리 파일 : 학과별_전임교원.csv 파일 사용
전처리 사항
1. 기준연도 2024 제거 (2010~2023의 데이터를 학습하는 프로젝트이기 때문)
2. 학생1인당 전임교원 수 계산 : 전임교원수/재학생 수
이유 : 학생입장에서는 같은 수학 지표더라도 더 의미 있을 수 있어서 추가
3. 2010~2023 공통학교 필터링 
3. 기준연도 별로 학교명 오름차순 정렬

```python
#  데이터 불러오기
import pandas as pd

# Colab 좌측 파일탐색기에 업로드한 경우, 파일명만 사용해도 됨
df = pd.read_csv('학과별_전임교원.csv', encoding='utf-8')
```

```python
#  기준연도 2024 제거
df = df[df['기준연도'] < 2024].copy()

#  학생1인당 전임교원 수 계산 (재학생 ÷ 전임교원) 후 소수점 첫째 자리 반올림
df['학생1인당전임교원'] = (df['재학생'] / df['전임교원']).round(1)

#  기준연도 오름차순 정렬
df = df.sort_values(by='기준연도').reset_index(drop=True)
```

```python
# 공통학교 필터링 및 정렬 (기준연도 → 학교명 순)
# 2023년 기준 학교명 목록 추출
schools_2023 = df[df['기준연도'] == 2023]['학교명'].unique()

# 전체 연도 데이터에서 2023년도에 존재한 학교만 남김
common_school_df = df[df['학교명'].isin(schools_2023)]

# 연도별 등장 횟수 계산 (학교명 기준)
year_counts = common_school_df.groupby('학교명')['기준연도'].nunique()

# 모든 연도(2010~2023: 14년)에 등장한 학교만 필터링
common_schools_all_years = year_counts[year_counts == 14].index

# 최종 공통학교 데이터셋 생성 후 정렬
df_common = common_school_df[common_school_df['학교명'].isin(common_schools_all_years)].copy()
df_common = df_common.sort_values(by=['기준연도', '학교명']).reset_index(drop=True)

# 확인
print(f"공통학교 수: {df_common['학교명'].nunique()}개")
df_common.head()
```

```python
# ✅ 공통학교 데이터 저장
df_common.to_csv('전처리_완료_공통학교_학과별_전임교원.csv', index=False, encoding='utf-8-sig')

# Colab 환경에서 직접 다운로드
from google.colab import files
files.download('전처리_완료_공통학교_학과별_전임교원.csv')
```
