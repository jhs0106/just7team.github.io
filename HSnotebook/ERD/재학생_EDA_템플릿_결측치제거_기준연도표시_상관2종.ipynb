```python
df = pd.read_csv("결측치제거_공통학교만_통합데이터_UTF8BOM.csv", encoding="utf-8-sig")
df.shape
```

```python
df.isnull().sum()
```

```python
numeric_cols = df.select_dtypes(include='number').columns.drop('기준연도', errors='ignore')
df[numeric_cols].hist(bins=20, figsize=(15, 10))
plt.suptitle("수치형 피처 분포", fontsize=16)
plt.show()
```

```python
corr = df[numeric_cols].corr()
plt.figure(figsize=(12, 8))
sns.heatmap(corr, annot=True, fmt=".2f", cmap='coolwarm')
plt.title("전체 상관관계 히트맵")
plt.show()
```

```python
# 중도탈락율과 다른 변수 간 상관관계
target = "중도탈락율"
target_corr = df[numeric_cols].corr()[[target]].drop(index=target).sort_values(by=target, ascending=False)
plt.figure(figsize=(6, len(target_corr) * 0.5))
sns.heatmap(target_corr, annot=True, fmt=".2f", cmap="coolwarm")
plt.title(f"{target}과 다른 변수 간 상관관계")
plt.show()
```

```python
for col in numeric_cols:
    if col != "중도탈락율":
        plt.figure()
        sns.scatterplot(data=df, x=col, y="중도탈락율", hue=df['기준연도'], palette="tab10")
        plt.title(f"중도탈락율 vs {col} (기준연도 표시)")
        plt.xlabel(col)
        plt.ylabel("중도탈락율")
        plt.legend(title="기준연도")
        plt.show()
```
